## Quick prediction example

```python
from preprocess import preprocess_pipeline
import joblib

model = joblib.load("../models/best_model.joblib")
vectorizer = joblib.load("../models/tfidf_vectorizer.joblib")

examples = [
    "Massive flood in Ahmedabad — roads submerged near Sardar Bridge.",
    "My exam results are a complete disaster lol"
]

for text in examples:
    cleaned = preprocess_pipeline(text)
    X = vectorizer.transform([cleaned])
    pred = model.predict(X)[0]
    proba = model.predict_proba(X)[0] if hasattr(model, "predict_proba") else [0, 0]
    label = "REAL DISASTER" if pred == 1 else "NOT A DISASTER"
    print(f"{text[:60]}... → {label} (conf={max(proba):.2f})")
```
